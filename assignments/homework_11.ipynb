def f(l):
    t=0
    for i in l:
        t+=i
    return t/len(l)
def g(l,m):
    return [x - m for x in l]
def h(l):
    m=f(l);d=g(l,m)
    s=0
    for x in d:s+=x*x
    return s/len(d)
