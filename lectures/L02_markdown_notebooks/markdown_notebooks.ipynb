## Why Markdown?

Markdown is a lightweight text format that converts to nicely formatted
documents. You'll see it in three places constantly:

- **Jupyter notebooks** — the non-code cells are Markdown.
- **GitHub** — every README, issue, and pull-request description.
- **Slidev** — the slide decks we use in class are Markdown.

It's worth one solid hour to internalize. After that, it's reflex.

### A few math tips

- Backslash-escape special LaTeX symbols: `\alpha`, `\beta`, `\sum`.

- Aligned equations use a `\begin{align*} ... \end{align*}` environment.

## The Jupyter cell model

A notebook is a sequence of **cells**. Each cell is either:

- a **Markdown cell** (prose and math, like this one), or
- a **code cell** (Python that runs in the kernel).

You're always in one of two modes:

- **Edit mode** (green border) — you're editing the contents of a cell.
- **Command mode** (blue border) — the cell is selected but not being
  edited. Keyboard shortcuts do notebook-level things.

Press `Esc` to go from edit to command mode. Press `Enter` to go back.


### Essential shortcuts

In command mode (`Esc` first):

| Keys | Does |
|------|------|
| `a` | Insert cell **above** |
| `b` | Insert cell **below** |
| `dd` | **Delete** the current cell |
| `m` | Convert cell to **Markdown** |
| `y` | Convert cell to **code** (Py**y**thon) |
| `Shift+Enter` | Run cell, select next |
| `Ctrl+Enter` | Run cell, stay put |
| `Alt+Enter` | Run cell, insert new cell below |
| `z` | Undo cell delete |

Memorize `a`, `b`, `dd`, `m`, and `Shift+Enter` first. The rest follow.


## The kernel

A **kernel** is the actual Python process that runs your code. When you
start a notebook, Jupyter spawns a kernel. When you run a cell, the code
goes to that kernel.

Important consequences:

- The kernel keeps state **between cells**. A variable defined in cell 3
  is still there in cell 17.
- **Restarting the kernel** wipes that state.
- **Closing the tab** does *not* always kill the kernel. Use
  Kernel → Shutdown when you're done.

Let's see this. Run the three cells below in order.


````python
x = 41
````

````python
x = x + 1
````

````python
print(x)  # 42
````

Now go back to the cell that says `x = x + 1` and run it again. And
again. What does the final cell print?


### The "hidden state" bug

This ability to re-run cells out of order is also the #1 source of
notebook bugs. Consider:

1. Cell A: `x = 1`
2. Cell B: `x = x + 1`
3. Cell C: `print(x)`

Run them top to bottom → prints `2`. Perfect.

Now edit cell A to say `x = 10`. If you re-run only A and C (skipping B)
it prints `10`. If you re-run A, B, C it prints `11`. If you re-run A, B,
B, C it prints `12`. A reader opening your notebook tomorrow has no way to
know which of those happened.

**Two habits that make this go away:**

1. Use **Kernel → Restart Kernel and Run All** before you commit,
   publish, or share a notebook. The output you see must be the output
   that a fresh kernel would produce.
2. Write cells so they're **idempotent** — you can re-run them in any
   order and get the same answer. Avoid `x = x + 1` style assignments at
   the top level.


## Running shell commands from a cell

You can run a shell command from inside a code cell with a leading `!`:


````python
!echo "hello from the shell"
````

You can capture the output into a Python variable with the same
syntax:


````python
files = !ls /
print(files[:3])
````

This is handy for quick tasks but don't overuse it — for anything
more than one command, write a proper shell script (week 3) or a Python
script.


## Magics

Lines starting with `%` are **line magics** (affect one line); lines
starting with `%%` are **cell magics** (affect the whole cell). A few
useful ones:


````python
%%time
total = 0
for i in range(10_000):
    total += i
````

````python
%whos
````

`%whos` shows the variables currently alive in the kernel. If your
notebook is behaving strangely, this is a good first thing to check.


## Exporting a notebook

From JupyterLab: **File → Save and Export Notebook As... → HTML / PDF /
Markdown**.

From the command line (inside your project):

```sh
uv run jupyter nbconvert --to html your_notebook.ipynb
uv run jupyter nbconvert --to pdf  your_notebook.ipynb   # needs LaTeX
```

For homework submissions, upload the `.ipynb` itself (not the exported
HTML) so we can re-run it.
