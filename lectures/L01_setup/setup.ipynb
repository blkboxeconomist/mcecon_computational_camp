## Why we care about tooling

Before we write a single line of economics code, we're going to spend an
hour on **tooling**. This is deliberate.

A huge fraction of the frustration new researchers feel with Python is not
actually about Python — it's about:

- *"Which Python am I running? Why is `pandas` not installed?"*
- *"My notebook worked yesterday and now it doesn't."*
- *"A collaborator ran my code and got a totally different answer."*

Every one of those problems is a **reproducibility** or **environment**
problem. A good toolchain makes them mostly go away.


## What we'll install today

| Tool | Why |
|------|-----|
| [`uv`](https://docs.astral.sh/uv/) | Manages Python versions, project envs, and dependencies. Replaces `pip`, `venv`, `pyenv`, and most of conda. |
| **Python** | Installed *through* `uv`, so every project can pin its own version. |
| **VS Code** *or* **Zed** | A text editor — pick either. Both have good Python and Jupyter support. |
| **JupyterLab** | The notebook interface you'll use for every lecture and most homework. |

If you already have Python, an editor, or Jupyter installed from a previous
class — great. We'll still do everything below so that your camp work lives
in a clean, reproducible environment.

## Step 1 — Install `uv`

`uv` is a single self-contained binary. You install it with one command.

**macOS and Linux** (in your terminal):

```sh
curl -LsSf https://astral.sh/uv/install.sh | sh
```

**Windows** (in PowerShell):

```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

After it finishes, close and reopen your terminal so the new `uv` command
shows up on your `PATH`.


### Verify `uv` is installed

In a fresh terminal, run:

```sh
uv --version
```

You should see something like `uv 0.5.x`. If instead you see
`command not found`, the installer's instructions at the end of its output
tell you which line to add to your shell config (`~/.zshrc` on macOS,
`~/.bashrc` on most Linux).


## Step 2 — Install Python through `uv`

Even if your machine already has a Python installed, we want `uv` to manage
it. That way every project can pin a specific version and we don't depend
on whatever the OS happens to ship.

```sh
uv python install 3.12
```

This downloads a standalone Python 3.12 build and puts it under `uv`'s
control. It does **not** touch your system Python.

Check what `uv` knows about:

```sh
uv python list --only-installed
```


## Step 3 — Create your first project

Let's make a fresh directory for a throwaway project.

```sh
mkdir ~/camp-sandbox
cd ~/camp-sandbox
uv init --python 3.12
```

`uv init` creates four files:

```
camp-sandbox/
├── .python-version     # pins Python to 3.12 for this project
├── main.py             # a tiny "hello world" script
├── pyproject.toml      # the project's metadata and dependency list
└── README.md
```

`pyproject.toml` is the important one. It's the single source of truth for
what this project is and what it needs.


### Running the starter script

```sh
uv run main.py
```

The first time you run `uv run ...` inside a project, `uv`:

1. Creates a hidden `.venv/` directory for this project's packages.
2. Installs the pinned Python into it.
3. Runs your command with that Python.

Every subsequent `uv run` is essentially instant — the env already exists.


## Step 4 — Add dependencies

`uv add <package>` does two things at once:

- Installs the package into this project's `.venv`.
- Writes the requirement into `pyproject.toml` so it's permanent.

For the camp, we need JupyterLab and `ipykernel`:

```sh
uv add jupyterlab ipykernel
```

Open `pyproject.toml` in any text editor and you should see new lines under
`dependencies = [...]`.


### About `uv.lock`

`uv` also writes a file called `uv.lock`. You shouldn't edit it by hand.

`uv.lock` records the **exact** version of every dependency (and every
transitive dependency) that was resolved. Commit it to Git. If you send
your project to a collaborator, `uv sync` on their machine will reproduce
*your* environment down to the patch version. No more
"it works on my machine".


## Step 5 — `uv run` is your friend

Any time you want to run a command inside the project's environment, prefix
it with `uv run`:

```sh
uv run python              # open a Python REPL with this project's deps
uv run python script.py    # run a script
uv run jupyter lab         # launch JupyterLab
uv run pytest              # (later) run tests
```

You don't need to "activate" a virtual environment. `uv run` handles that
for you.


## Step 6 — Install a text editor

A text editor is where you'll actually write code. For the camp, pick
one of these two:

- **[VS Code](https://code.visualstudio.com/)** — the most common choice.
  Huge extension ecosystem, integrates with every AI assistant, works
  identically on macOS, Linux, and Windows.
- **[Zed](https://zed.dev/)** — newer, much faster, simpler UI, great
  Python + LSP support out of the box. Good AI integration too.

Both are free and cross-platform. Either is a fine choice for camp and
for the follow-on courses. If you have no preference, VS Code is the
safer default because it's what most of your classmates and collaborators
will use; if you value a lean, fast editor, try Zed.

### Other editors you might hear about

- **Cursor** — a VS Code fork with AI features built in. Week 9.
- **PyCharm** — heavier IDE; some people love it. Fine if you already use it.
- **Vim / Neovim / Helix** — great once you've invested time. Not for
  week 1.

### Recommended editor setup

**If you picked VS Code:** open the Extensions panel
(`Ctrl+Shift+X` / `Cmd+Shift+X`) and install:

| Extension | Publisher |
|-----------|-----------|
| **Python** | Microsoft |
| **Jupyter** | Microsoft |
| **Ruff** | Astral Software |

The Python extension gives you IntelliSense, linting, and the test runner.
The Jupyter extension lets you open `.ipynb` files directly in VS Code.
The Ruff extension formats on save (we'll see why in week 11).

**If you picked Zed:** most of what you need is already built in.
Zed ships with Python LSP support (Pyright) and tree-sitter syntax out of
the box. Two things worth setting up:

- **Ruff** — Zed can use Ruff as a formatter/linter. In your
  `~/.config/zed/settings.json`, add a `"languages": { "Python": { ... } }`
  block pointing the formatter at Ruff (Zed's Python docs have the
  current recipe).
- **Notebooks** — Zed has native `.ipynb` support; opening an `.ipynb`
  file gives you a notebook UI with cells and a kernel picker.

### A tiny bit of real code

Notebooks aren't just for printing strings. Run this one too — it's a toy
simulation of compound interest. We'll see *much* more of this flavor over
the next 11 weeks.


````python
principal = 1000.0
rate = 0.05
years = 10

balance = principal
for year in range(1, years + 1):
    balance = balance * (1 + rate)
    print(f"year {year:2d}: ${balance:,.2f}")
````

## Common gotchas

**"`uv` command not found" after install.**
Close your terminal and open a new one. If that doesn't work, follow the
PATH hint the installer printed.

**"Which Python am I using?"**
Inside a project, always `uv run python`. Outside a project, the answer is
"it depends, and you shouldn't care." Make a project.

**"Jupyter can't find my packages."**
You almost certainly ran `jupyter lab` from outside the project directory,
or without `uv run`. Both matter: `cd` into the project *and* prefix with
`uv run`.

**"I already have conda / pyenv / pipx installed."**
They can coexist with `uv`, but for the camp use `uv` exclusively to avoid
confusion. If something breaks and you're unsure which tool is involved,
ask.

One slide's worth of commands you'll use constantly:

```sh

```
