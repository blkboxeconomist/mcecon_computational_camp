## Indentation matters

Python uses **indentation** to show block structure — the code "inside"
an `if` or a loop. The standard is **four spaces**. Don't mix tabs and
spaces. Your editor should handle this; if it doesn't, change editors.


## `if`, `elif`, `else`

Structure:

```python
if condition:
    # runs when condition is truthy
elif other_condition:
    # runs when the first was false and this is truthy
else:
    # runs otherwise
```

You must have `if`. `elif` and `else` are optional, and you can have
any number of `elif` branches.


````python
x = 17

if x < 0:
    print("negative")
elif x == 0:
    print("zero")
elif x < 10:
    print("small positive")
else:
    print("large")
````

### Conditional expressions (ternary)

For simple "one value or another" cases, use the expression form:

```python
y = a if condition else b
```


````python
age = 20
status = "adult" if age >= 18 else "minor"
print(status)
````

## `while` loops

```python
while condition:
    # runs repeatedly while condition is truthy
```

Whoever writes the loop is responsible for making the condition
eventually become false. Otherwise: infinite loop (hit `Stop` / kernel
interrupt).


````python
# Count down from 5
n = 5
while n > 0:
    print(n)
    n = n - 1  # or: n -= 1
print("liftoff")
````

### A small simulation

Consumption under a simple rule: keep consuming 10% of your wealth per
period until it falls below a threshold.


````python
wealth = 1000.0
period = 0

while wealth > 100:
    consumption = 0.10 * wealth
    wealth = wealth - consumption
    period += 1
    print(f"period {period}: consumed {consumption:7.2f}, wealth = {wealth:7.2f}")

print(f"fell below threshold after {period} periods")
````

## `for` loops

`for` iterates over any **iterable** — ranges, strings, lists, files,
and much more. We haven't seen lists yet (week 7), but `range` is
enough for now.


### `range`

`range(n)` gives the integers `0, 1, ..., n-1`. It does **not** include
`n`.

| Call | Produces |
|------|----------|
| `range(5)` | `0, 1, 2, 3, 4` |
| `range(2, 5)` | `2, 3, 4` |
| `range(0, 10, 2)` | `0, 2, 4, 6, 8` |
| `range(5, 0, -1)` | `5, 4, 3, 2, 1` |


````python
for i in range(5):
    print(i)
````

````python
for i in range(5, 10, 2):
    print(i)
````

### Iterating over a string


````python
for letter in "camp":
    print(letter.upper())
````

### Accumulator pattern

Loops often accumulate a running total. Start outside the loop, update
inside.


````python
# Sum 1 + 2 + ... + 100
total = 0
for i in range(1, 101):
    total = total + i
print(total)  # 5050

# Factorial of 10
result = 1
for i in range(1, 11):
    result *= i
print(result)  # 3628800
````

## `break` and `continue`

- `break` immediately exits the innermost loop.
- `continue` skips to the next iteration of the innermost loop.


````python
# First multiple of 7 that's greater than 50
for i in range(100):
    if i > 50 and i % 7 == 0:
        print("found", i)
        break
````

````python
# Print only the odd numbers from 0..9
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
````

## `enumerate`

When you need both the index and the value:


````python
for index, letter in enumerate("camp"):
    print(index, letter)
````

Start the index from 1 (handy for human-readable output):


````python
for n, letter in enumerate("camp", start=1):
    print(f"{n}. {letter}")
````

## Nested loops

Loops can nest. Each iteration of the outer loop runs the entire inner
loop.


````python
for i in range(3):
    for j in range(3):
        print(f"({i}, {j})", end="  ")
    print()
````

## A mini example: find primes up to 30

Not the most efficient algorithm — but a good exercise in nested
control flow.


````python
for n in range(2, 31):
    is_prime = True
    for d in range(2, n):
        if n % d == 0:
            is_prime = False
            break
    if is_prime:
        print(n, end=" ")
print()
````

## A mini example: FizzBuzz

An interview classic. Print 1..20, replacing multiples of 3 with "Fizz",
multiples of 5 with "Buzz", and multiples of both with "FizzBuzz".

*Before running, predict what line 13 (for `n = 13`) prints.*


````python
for n in range(1, 21):
    if n % 15 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
````

**Why did we check 15 first?** If we checked `n % 3` first, `15`
would match and print `"Fizz"`, never reaching the `FizzBuzz` case. Order
of `if/elif` matters.


## `while` vs `for`: which to use?

- Use **`for`** when you know *how many times* you'll iterate or *what
  you're iterating over* (a range, a string, later a list).
- Use **`while`** when you're iterating until some **condition** is met
  and the count isn't known in advance (e.g., "until the error is small
  enough", "until the user quits").

Most loops end up being `for` loops in practice.


## Recap

- `if`/`elif`/`else`, indented with four spaces

- `for x in iterable:` for the common case; use `range(...)` for integers
- `break` / `continue` to escape or skip
- `enumerate(...)` when you need an index alongside a value
