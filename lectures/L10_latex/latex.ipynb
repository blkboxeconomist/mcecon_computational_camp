## Why LaTeX

Economics papers are typeset in LaTeX. Journal submissions require it.
Tenured faculty still fight about it. Learning it well now pays you back
every year.

The core value: you write content (words, equations, tables) and LaTeX
handles layout. Change the document class and a draft becomes a journal
submission with no manual reformatting.


## Installing LaTeX

Two good options for the camp:

1. **Local** — [`tectonic`](https://tectonic-typesetting.github.io/) is
   the simplest: one binary, automatically downloads packages.

   ```sh
   # macOS
   brew install tectonic

   # Linux
   cargo install tectonic      # or your distro's package
   ```

   Use with: `tectonic main.tex`.

2. **Overleaf** — web-based, no install. Free tier is fine for the
   camp. Sign in with your Rice email for some extra features.

A heavier local alternative is **TeX Live** + `latexmk`. Fine if you
already have it. Otherwise, start with `tectonic` or Overleaf.


## A minimal document

Save this as `hello.tex`:

```tex
\documentclass[11pt]{article}
\usepackage[a4paper, margin=1in]{geometry}
\usepackage{amsmath, amssymb}

\title{My first LaTeX document}
\author{Your Name}
\date{\today}

\begin{document}
\maketitle

Hello, world.

\end{document}
```

Compile with `tectonic hello.tex`. You get `hello.pdf`.


## Document structure

A typical short paper:

```tex
\documentclass[11pt]{article}

\usepackage[a4paper, margin=1in]{geometry}
\usepackage{amsmath, amssymb, amsthm}
\usepackage{graphicx}
\usepackage{booktabs}
\usepackage[style=authoryear]{biblatex}
\addbibresource{refs.bib}

\title{...}
\author{...}
\date{\today}

\begin{document}
\maketitle
\begin{abstract}
  ...
\end{abstract}

\section{Introduction}
...

\section{Model}
...

\section{Results}
...

\section{Conclusion}
...

\printbibliography
\end{document}
```

Packages you'll use constantly:

- `amsmath`, `amssymb` — better math
- `graphicx` — include images
- `booktabs` — professional tables
- `biblatex` — bibliographies
- `hyperref` — clickable links


## Math

### Inline

Wrap with `$...$`:

```tex
The budget constraint is $c + s = y$.
```

### Display

Unnumbered, with `\[...\]`:

```tex
\[
V(k) = \max_{c,\, k'}\; u(c) + \beta V(k')
\]
```

Numbered, with an `equation` environment:

```tex
\begin{equation}
  V(k) = \max_{c,\, k'}\; u(c) + \beta V(k')
  \label{eq:bellman}
\end{equation}
```

Reference it later: `Equation~\eqref{eq:bellman} shows ...`


### Aligned math

```tex
\begin{align}
  u'(c_t) &= \beta (1 + r) u'(c_{t+1}) \\
  c_t + s_t &= y_t + (1 + r) s_{t-1}
\end{align}
```

- `&` — alignment point (usually just before the `=`).
- `\\` — end of line.

Use `align*` for unnumbered aligned equations.


### Common math snippets

```
\alpha, \beta, \gamma, \dots, \omega
\Alpha, \Beta, \Gamma, \Delta, \Omega         % uppercase Greek
\sum_{i=1}^N  \prod_{i=1}^N
\int_0^\infty  \int_\Omega
\mathbb{R}, \mathbb{E}, \mathbb{P}
\mathcal{L}, \mathcal{F}
\frac{a}{b}  \sqrt{x}  x^{2}  x_{t+1}
\left( ... \right)   % auto-sized parens
\lVert x \rVert      % norm
\approx, \neq, \leq, \geq, \to, \infty
```

You don't need to memorize all of this. Keep `amsmath` documentation or
Detexify (a symbol search tool) open while you write.


## Tables

Use `booktabs` for professional-looking tables. Never use vertical
rules.

```tex
\begin{table}[h]
  \centering
  \caption{GDP growth in selected countries.}
  \label{tab:gdp}
  \begin{tabular}{lrr}
    \toprule
    Country & 2022 & 2023 \\
    \midrule
    USA     & 2.1  & 2.5 \\
    China   & 3.0  & 5.2 \\
    Germany & 1.8  & $-0.3$ \\
    \bottomrule
  \end{tabular}
\end{table}
```

`l`, `c`, `r` in the `{lrr}` describe column alignment. `&` separates
columns; `\\` ends a row.


## Figures

```tex
\begin{figure}[h]
  \centering
  \includegraphics[width=0.7\linewidth]{figures/plot.pdf}
  \caption{Year-over-year GDP growth.}
  \label{fig:gdp}
\end{figure}
```

Prefer `.pdf` or `.png` figures. `.pdf` scales without pixelation —
matplotlib can save directly to PDF: `plt.savefig("plot.pdf")`.

Reference with `Figure~\ref{fig:gdp}`.


## Bibliography

Keep citations in a `.bib` file:

```bibtex
% refs.bib
@article{acemoglu2001colonial,
  author = {Acemoglu, Daron and Johnson, Simon and Robinson, James A.},
  title  = {The Colonial Origins of Comparative Development},
  journal = {American Economic Review},
  volume = {91},
  number = {5},
  year   = {2001},
  pages  = {1369--1401},
}
```

Cite it in the `.tex`:

```tex
The seminal study \parencite{acemoglu2001colonial} showed ...
```

- `\cite{key}` — plain
- `\parencite{key}` — parenthetical (Author, Year)
- `\textcite{key}` — textual: "Author (Year)"

End of document:

```tex
\printbibliography
```

Most journals and services will accept a `.bib` file directly from
Google Scholar's "Cite" button.


## `~`, `--`, `---`, `%`, `&`, `#`, `_`, `$`

LaTeX's special characters:

| Character | What to type |
|-----------|--------------|
| `%` | `\%` |
| `&` | `\&` |
| `$` | `\$` |
| `#` | `\#` |
| `_` | `\_` |
| `~` (non-breaking space) | `~` (use between name + number, like `Figure~1`) |
| `--` | `--` (for en-dashes, page ranges) |
| `---` | `---` (em-dashes) |

Forgetting `\%` in a table will give you a baffling error. The shape of
the error will be "Missing $ inserted". Remember the special-characters
rule.


## Workflow

1. Write in your favorite editor (VS Code has a great LaTeX Workshop
   extension; Zed has built-in LaTeX syntax + LSP via `texlab`).
2. Save. Compile (`tectonic main.tex` or click build in Overleaf).
3. View the PDF.
4. Iterate.

For anything longer than a page, **commit frequently with Git**. LaTeX
is text, so diffs are meaningful.

## Debugging LaTeX errors

A few survival tips:

- The error message is usually unhelpful. Scroll up in the output to
  find the **first** thing that went wrong.
- `! Undefined control sequence.` means LaTeX doesn't know a command —
  typo, or missing `\usepackage`.
- `! Missing $ inserted.` means you used a special character in text
  mode. Find the `%`, `&`, `#`, or `_` you forgot to escape.
- Cross-references (`\ref`, `\cite`) often need **two** compilations —
  the first writes the labels, the second resolves them.

AI assistants are great at translating LaTeX errors into plain English.


## Recap

- LaTeX separates content from layout. One package of styling rules
  becomes another.
- Minimal template: `\documentclass`, some `\usepackage` lines,
  `\begin{document}...\end{document}`.
- Math lives in `$...$`, `\[...\]`, `equation`, or `align`.
- `booktabs` for tables, `graphicx` for figures, `biblatex` for
  bibliography.
- Compile with `tectonic` or Overleaf.
