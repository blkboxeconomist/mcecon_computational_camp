## Why four different types?

All four hold multiple values, but they differ on axes that matter:

| Type | Ordered? | Mutable? | Unique? | Lookup by |
|------|----------|----------|---------|-----------|
| `list` | yes | yes | no | index |
| `tuple` | yes | no | no | index |
| `set` | no | yes | **yes** | membership |
| `dict` | yes* | yes | (keys unique) | **key** |

*Since Python 3.7, dicts preserve insertion order.

Memorize this table. Picking the right type makes code simpler.


## Lists

A list is an ordered, mutable sequence. Write one with `[...]`:


````python
gdp = [1000, 1050, 1103, 1158]
print(gdp)
print(type(gdp))
````

### Indexing and slicing

Same indexing rules as strings — 0-based, negative from the end, and
`[start:stop:step]` slicing.


````python
print(gdp[0])  # first
print(gdp[-1])  # last
print(gdp[1:3])  # [1050, 1103]
print(gdp[::-1])  # reversed copy
print(len(gdp))
````

### Mutation

Unlike strings, lists can be changed in place:


````python
gdp = [1000, 1050, 1103, 1158]

gdp[0] = 999  # replace
gdp.append(1210)  # add to end
gdp.insert(0, 950)  # insert at index 0
popped = gdp.pop()  # remove and return last
gdp.remove(1050)  # remove by value (first occurrence)

print(gdp, "popped:", popped)
````

### List as a stack / queue (informally)


````python
stack = []
stack.append(1)
stack.append(2)
stack.append(3)
print(stack.pop())  # 3
print(stack.pop())  # 2
````

### Iterating over a list


````python
for value in gdp:
    print(value)
````

````python
# With an index
for i, value in enumerate(gdp):
    print(f"period {i}: {value}")
````

### Useful list functions

```
len(lst)         length
sum(lst)         sum of numbers
max(lst), min(lst)
sorted(lst)      returns a sorted copy
lst.sort()       sorts in place (returns None!)
lst.count(x)     how many times x appears
x in lst         membership (scans, O(n))
```


````python
nums = [3, 1, 4, 1, 5, 9, 2, 6]
print(sum(nums), max(nums), min(nums))
print(sorted(nums))
print(sorted(nums, reverse=True))
print(nums.count(1))
````

## Tuples

A tuple is an ordered, **immutable** sequence. Written with `(...)` (or
no brackets at all — just commas):


````python
point = (3.0, 4.0)
point2 = 3.0, 4.0  # also a tuple
print(point == point2)
print(point[0])
# point[0] = 9        # TypeError: tuples are immutable
````

Tuples are the right choice when the set of values is fixed and
meaningful *as a whole*: (x, y) coordinates, (name, age) pairs,
function-return-multiples.


### Tuple unpacking

Assignment to a tuple splits the values out:


````python
coords = (3.0, 4.0)
x, y = coords
print(x, y)

name, age = "Chase", 36
print(name, age)

# Swap two variables in one line — classic
a, b = 1, 2
a, b = b, a
print(a, b)
````

## Dictionaries

A dict maps **keys** to **values**. Written with `{key: value, ...}`:


````python
gdp_by_country = {
    "USA": 25_000,
    "CHN": 17_700,
    "DEU": 4_200,
}
print(gdp_by_country)
print(gdp_by_country["USA"])
````

### Adding, updating, deleting


````python
gdp_by_country["JPN"] = 4_300  # add
gdp_by_country["USA"] = 26_000  # update
del gdp_by_country["DEU"]  # delete
print(gdp_by_country)
````

### Safely looking up a key

`d[key]` raises `KeyError` if the key is missing. `d.get(key, default)`
returns `default` instead.


````python
print(gdp_by_country.get("USA", "unknown"))
print(gdp_by_country.get("FRA", "unknown"))
````

### Iterating over a dict

Three flavors:


````python
# Keys
for country in gdp_by_country:
    print(country)
````

````python
# Values
for gdp in gdp_by_country.values():
    print(gdp)
````

````python
# Both
for country, gdp in gdp_by_country.items():
    print(f"{country}: ${gdp:,}M")
````

### When to reach for a dict

Any time you'd otherwise think "I want to look something up by a label
or ID". Counting occurrences. Grouping. Config.


## Sets

A set is an unordered collection of **unique** values. Written with
`{a, b, c}` (empty set is `set()`, not `{}` — that's a dict!).


````python
countries = {"USA", "USA", "CHN", "DEU"}
print(countries)  # duplicate USA collapsed
print("USA" in countries)  # fast membership
````

### Set operations

```
a | b        union
a & b        intersection
a - b        difference
a ^ b        symmetric difference (in either but not both)
```


````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)
print(a & b)
print(a - b)
print(a ^ b)
````

### When to reach for a set

- Deduplicating (`unique = set(lst)`)
- Fast membership tests on a big collection
- "Which elements are in A but not B?"


## Comprehensions

Comprehensions build a new list / set / dict in one expression. This is
one of the most idiomatic patterns in Python — worth a careful look.

**List comprehension:**

```python
[expr for x in iterable]
[expr for x in iterable if condition]
```


````python
squares = [n * n for n in range(10)]
print(squares)

even_squares = [n * n for n in range(10) if n % 2 == 0]
print(even_squares)
````

**Set comprehension:**


````python
letters = {ch.lower() for ch in "Economics" if ch.isalpha()}
print(letters)
````

**Dict comprehension:**


````python
gdp_by_country = {"USA": 25_000, "CHN": 17_700, "DEU": 4_200}
usd_to_thousand = {k: v / 1_000 for k, v in gdp_by_country.items()}
print(usd_to_thousand)
````

### Comprehension style

Comprehensions are great when the body is one expression. If you need
multiple statements inside the loop, write a regular `for` loop. Don't
force everything into a one-liner.


## Zipping and unpacking together


````python
years = [2020, 2021, 2022, 2023]
gdp = [1000, 1050, 1103, 1158]

# Pair them up
for year, value in zip(years, gdp):
    print(year, value)

# Build a dict from two parallel lists
gdp_by_year = dict(zip(years, gdp))
print(gdp_by_year)
````

## Picking the right structure

A quick mental decision tree:

- Do I need to look things up by a label? → **dict**
- Do I need every element to be unique? → **set**
- Is this a fixed small group of related values? → **tuple**
- Ordered, growable, indexable? → **list**


## Recap

- `list`, `tuple`, `set`, `dict` — know the trade-offs.
- Mutation vs. immutability is the big axis.
- Comprehensions build new collections in one expression.
- `zip` + `dict(zip(...))` is a very common pair.
