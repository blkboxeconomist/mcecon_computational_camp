## Variables and assignment

A **variable** is a name bound to a value. You create one with `=`:


````python
x = 42
name = "Chase"
tuition = 62346.0
````

No type declaration — Python figures out the type from the value.
You can rebind a variable to a different type:


````python
y = 10  # int
y = "ten"  # now a string — legal, though rarely what you want
print(y)
````

Multiple assignment:


````python
a, b = 1, 2
print(a, b)
````

## Core types

Four types carry most of the weight early on:

| Type | Examples | Literal form |
|------|----------|--------------|
| `int` | `1`, `-17`, `1_000_000` | no decimal point |
| `float` | `3.14`, `1.0`, `2e-3` | has `.` or `e` |
| `str` | `"hello"`, `'hi'` | quotes |
| `bool` | `True`, `False` | capitalized |

Underscores in numbers are just for readability — Python ignores them.


````python
million = 1_000_000
print(million, type(million))
````

`type(x)` tells you what type `x` is. Useful for debugging.


````python
for value in (42, 3.14, "hi", True):
    print(value, "->", type(value).__name__)
````

## Numbers: arithmetic

```
+   addition              7 + 2   ==  9
-   subtraction           7 - 2   ==  5
*   multiplication        7 * 2   == 14
/   true division         7 / 2   ==  3.5
//  floor division        7 // 2  ==  3
%   modulo (remainder)    7 % 2   ==  1
**  exponentiation        7 ** 2  == 49
```


````python
print(7 / 2)  # 3.5 (always float)
print(7 // 2)  # 3   (integer divide, floors toward -inf)
print(-7 // 2)  # -4  (watch out)
print(7 % 2)  # 1
print(2**10)  # 1024
````

**Floats are not exact.** This will bite you eventually:


````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

`0.1 + 0.2` is not exactly `0.3` in binary floating point. When you
compare floats, compare with a tolerance:


````python
abs((0.1 + 0.2) - 0.3) < 1e-9
````

````python
first = "Hello"
second = "camp"
print(first + ", " + second + "!")
print("-" * 20)
````

### f-strings

The single most useful string feature. Prefix with `f` and embed
expressions in `{...}`:


````python
name = "Chase"
tuition = 62346.0
print(f"Hi {name}, your tuition is ${tuition:,.2f}.")
````

Format specifiers inside `{...}`:

| Spec | Result for 1234.5 |
|------|---------------------|
| `{x}` | `1234.5` |
| `{x:.2f}` | `1234.50` |
| `{x:,.2f}` | `1,234.50` |
| `{x:10.2f}` | `"   1234.50"` (right-padded to 10) |
| `{x:>10}` | right-aligned |
| `{x:<10}` | left-aligned |
| `{x:^10}` | centered |


````python
pi = 3.141592653589793
print(f"pi      = {pi}")
print(f"pi (6f) = {pi:.6f}")
print(f"pi (2f) = {pi:.2f}")
````

### Useful string methods


````python
s = "  Hello, Camp!  "

print(s.strip())  # "Hello, Camp!"
print(s.strip().lower())  # "hello, camp!"
print(s.strip().upper())  # "HELLO, CAMP!"
print(s.replace("Camp", "World"))
print(s.strip().split(", "))  # ['Hello', 'Camp!']
print(",".join(["a", "b", "c"]))
````

````python
email = "chase@rice.edu"
print(email.startswith("chase"))
print(email.endswith(".edu"))
print("rice" in email)
````

### String slicing

Every string has indices — 0-based, and negative indices count from the
end.


````python
s = "economics"
print(s[0])  # 'e'
print(s[-1])  # 's'
print(s[0:3])  # 'eco'
print(s[:3])  # 'eco' (omit start → 0)
print(s[3:])  # 'nomics' (omit end → len(s))
print(s[::-1])  # 'scimonoce' (reversed!)
````

The general slice is `s[start:stop:step]`. `stop` is exclusive —
this is *the* Python off-by-one pattern and it applies everywhere.


## Booleans and comparisons

Six comparison operators produce `bool`s:

```
==  equal            x == y
!=  not equal        x != y
<   less than        x < y
<=  less or equal    x <= y
>   greater than     x > y
>=  greater or equal x >= y
```


````python
print(3 < 5)
print("a" < "b")  # lexicographic
print("apple" == "apple")
````

### Boolean operators

```
and   both true
or    either true
not   negate
```


````python
age = 24
income = 55_000
print(age >= 18 and income >= 50_000)
print(not (age < 18))
````

### Truthiness

Many non-bool values are treated as truthy/falsy:

- `False`, `None`, `0`, `0.0`, `""`, `[]`, `{}`, `set()` — **falsy**
- Essentially everything else — **truthy**

So:


````python
if "":
    print("empty string is truthy")
else:
    print("empty string is falsy")
````

````python
for value in (0, 0.0, "", [], {}, None, "hi", 42):
    print(repr(value), "->", bool(value))
````

### `==` vs `is`

- `==` asks "are these values equal?"
- `is` asks "are these the *same object* in memory?"

99% of the time you want `==`. Use `is` only for `None`, `True`, `False`,
and other singletons.


````python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)  # True: same value
print(a is b)  # False: different objects
print(a is a)  # True
````

````python
x = None
print(x is None)  # idiomatic
````

## Type conversions


````python
# string -> number
print(int("42"))  # 42
print(float("3.14"))  # 3.14

# number -> string
print(str(42))  # "42"

# number -> bool
print(bool(0), bool(1))

# list of characters -> joined string
print("".join(["h", "i"]))
````

````python
# common pitfall
x = "7"
y = 3
# print(x + y)   # TypeError! can't add str + int
print(int(x) + y)
````
